## Matplotlib practice before the assignment charts

The next cells demonstrate the basic pattern for making a chart. These examples use small practice lists, not the answers to the assignment.

```python
# Create practice values for the horizontal axis.
practice_days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

# Create practice values for the vertical axis.
practice_hours = [1, 2, 2, 3, 4]

# Set the size of the chart in inches.
plt.figure(figsize=(8, 4))

# Draw a line chart using the two practice lists.
plt.plot(practice_days, practice_hours, color="teal", marker="o")

# Add a title and labels that explain the axes.
plt.title("Practice Study Time")
plt.xlabel("Day")
plt.ylabel("Hours")

# Add gridlines to make the values easier to compare.
plt.grid(alpha=0.3)

# Display the completed chart.
plt.show()
```
